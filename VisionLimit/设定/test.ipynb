```python
def find_max_index(lst):
    max_val = lst[0]
    for value in lst:
        if value > max_val:
            max_val = value
    return lst.index(max_val)


def find_peak_coordinate(grid):
    max_value_list = []
    for lst in grid:
        max_index = find_max_index(lst)
        max_value = lst[max_index]
        max_value_list.append(max_value)
    
    row_index = find_max_index(max_value_list)
    col_index = find_max_index(grid[row_index])
    return [row_index, col_index]
```

```python
# test
elevations = [
    [12, 15, 20],
    [18, 33, 24],
    [10, 29, 31]
]

print(find_peak_coordinate(elevations)) 
```
